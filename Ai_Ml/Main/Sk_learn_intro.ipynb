```python
import pandas as pd
```

***Getting Ready data***

```python
url = "https://raw.githubusercontent.com/mrdbourke/zero-to-mastery-ml/master/data/heart-disease.csv"
df = pd.read_csv(url)
df.head()
```

```python
# create x

x = df.drop("target", axis=1)
x
```

```python
# create y

y = df["target"]
y
```

***Choosing ML Model***

```python
from sklearn.ensemble import RandomForestClassifier
clf = RandomForestClassifier()
clf.get_params()
```

***Fitting Model***

```python
from sklearn.model_selection import train_test_split
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.3)
clf.fit(x_train, y_train)
```

***Evaluate Model***

```python
y_predicted = clf.predict(x_test)
y_predicted
```

```python
# Evaluation

clf.score(x_train, y_train)
```

```python
clf.score(x_test, y_test)
```

***Improve Model***

```python
for i in range(10,210,10):
  print(f"Running Model with {i} estimators.")
  clf = RandomForestClassifier(n_estimators=i).fit(x_train, y_train)
  print(f"Accuracy is : {(clf.score(x_test, y_test))*100}")

#  we find 180 gives us good accuracy
```

***Saving Model***

```python
import pickle
pickle.dump(clf,open("Heart_Disease_Predictor.pkl","wb"))
```

```python
load_model = pickle.load(open("/content/Heart_Disease_Predictor.pkl", "rb"))
load_model.score(x_test,y_test)
```

```python
import pandas as pd
Missing_phone = pd.read_csv("https://raw.githubusercontent.com/MehmoodCoder/Ai_Ml_Course/main/Ai_Ml/Data/mising_values_phone.csv")
Missing_phone
```

```python
Missing_phone.isna().sum()
```

```python
# we never apply model on data with NaN
```

```python

# fill missing values NaN

Missing_phone["brand"] = Missing_phone["brand"].fillna("missing")
Missing_phone["model"] = Missing_phone["model"].fillna("missing")
Missing_phone["condition"] = Missing_phone["condition"].fillna("missing")

Missing_phone["storage_gb"] = Missing_phone["storage_gb"].mean()
Missing_phone["ram_gb"] = Missing_phone["ram_gb"].mean()

# drop price

Missing_phone.dropna(inplace = True)
```

```python
Missing_phone.isna().sum()
```

```python

# method 2

Missing_phone_2 = pd.read_csv("https://raw.githubusercontent.com/MehmoodCoder/Ai_Ml_Course/main/Ai_Ml/Data/mising_values_phone.csv")
Missing_phone_2
```

```python
Missing_phone_2.isna().sum()
```

```python


Missing_phone_2 = Missing_phone_2.dropna(subset = ["price_pkr"])

# it remove complete row in which price_pkr contains NaN value


Missing_phone_2.isna().sum()
```

```python
Missing_phone_2
```

```python
# split data

X = Missing_phone_2.drop("price_pkr", axis = 1)
Y = Missing_phone_2["price_pkr"]
```

```python
X
```

```python
Y
```

```python
# imputation
# filling values with sklearn

from sklearn.impute import SimpleImputer
from sklearn.compose import ColumnTransformer

# fill categorical values with missing and numeric with mean
```

```python


Categorical_features = SimpleImputer(strategy = "constant", fill_value = "missing")
Numerical_features = SimpleImputer(strategy = "mean")
```

```python
# define columns

cat_cols = ["brand","model","condition"]
num_cols = ["storage_gb","ram_gb"]


Imputer = ColumnTransformer([
    ("Categorical_feature", Categorical_features,  cat_cols),
    ("Numeric_feature", Numerical_features, num_cols)
])
```

```python



Filled_X = Imputer.fit_transform(X)
Filled_X
```

```python


Phone_Data = pd.DataFrame(Filled_X,  columns = ["brand","model","condition","storage","ram"])
```
