#Matplotlib

```python
%matplotlib inline

import matplotlib.pyplot as plt
import pandas as pd
import numpy as np
```

```python
plt.plot()
plt.show()
```

```python
a1 = np.random.randint(10,size=(5,3))
a2 = np.random.randint(10,size=(5,3))

plt.plot(a1,a2)
plt.show()
```

Methods to plot

```python
# Method 1

fig = plt.figure()
axis = fig.add_subplot()
axis.plot(a1,a2)
plt.show()

```

```python
# Method 2

fig2 = plt.figure()
sub_plot2 = fig2.add_axes([1,1,1,1])
sub_plot2.plot(a1,a2)
plt.show()

```

```python
# Method 3

fig3,sub_plot3 = plt.subplots()
sub_plot3.plot(a1,a2)
# Add features to plot

sub_plot3.set(title="Plot",xlabel="X",ylabel="Y")
plt.show()
```

```python
# Exercise

# Speed Distance Graph

speed = [0,10,20,40,60,100,150]
distance = [0,100,200,400,600,1000,1500]

canvus,axis = plt.subplots()
axis.plot(speed,distance)
axis.set(title="Speed Distance Graph",xlabel="Speed",ylabel="Distance")
plt.show()
```

```python
# Exercise

# Load And Extension

load = [0,10,20,40,60,100,150]
extension = [0,100,200,400,600,1000,1500]

canvus,axis = plt.subplots()
axis.plot(load,extension)
axis.set(title="Load Extension Graph",xlabel="Load",ylabel="Extension")
plt.show()

```

# ***One figure with multiple plots***

```python
figure,((sub_plot1,sub_plot2),(sub_plot3,sub_plot4)) = plt.subplots(ncols=2,nrows=2,figsize = (10,10))

sub_plot1.plot(speed,distance)
sub_plot1.set(title="Speed Distance Graph",xlabel="Speed",ylabel="Distance")

sub_plot2.plot(load,extension);
sub_plot2.set(title="Load Extension Graph",xlabel="Load",ylabel="Extension");

sub_plot3.plot(speed,distance)
sub_plot3.set(title="Speed Distance Graph",xlabel="Speed",ylabel="Distance")

sub_plot4.plot(load,extension);
sub_plot4.set(title="Load Extension Graph",xlabel="Load",ylabel="Extension");
```

Types of ***Plot***

```python
# Line Plot

a1 = np.random.randint(10,size=(5,3))
a2 = np.random.randint(10,size=(5,3))

plt.plot(a1,a2); # add semicolon to remove under address
```

```python
# Scatter Plot

plt.scatter(a1,a2);
```

```python
# bar Plot

shop = {
    "item1":20,
    "item2":30,
    "item3":40,
    "item4":50,
    "item5":60
}

plt.bar(shop.keys(),shop.values());# vertically
```

```python
# bar Plot

shop = {
    "item1":20,
    "item2":30,
    "item3":40,
    "item4":50,
    "item5":60
}

# if you want to change it horizontally

plt.barh(shop.keys(),shop.values());
```

```python
# Histogram Plot
x = [1,2,3,4,8,9,99,90]
plt.hist(x);
```

```python
y = np.random.randn(10000)
plt.hist(y);
```

```python
figure,((sub_plot1,sub_plot2),(sub_plot3,sub_plot4)) = plt.subplots(ncols=2,nrows=2,figsize = (10,10))

sub_plot1.plot(speed,distance)
sub_plot1.set(title="Speed Distance Graph",xlabel="Speed",ylabel="Distance")

sub_plot2.bar(shop.keys(),shop.values());
sub_plot2.set(title="Bar",xlabel="X",ylabel="Y");

sub_plot3.hist(y)
sub_plot3.set(title="Histogram",xlabel="X",ylabel="Y")

sub_plot4.scatter(load,extension);
sub_plot4.set(title="Load Extension Graph",xlabel="Load",ylabel="Extension");
```
