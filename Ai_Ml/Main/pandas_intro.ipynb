```python
# Import Mising values file

import pandas as pd

phone_mv = pd.read_csv("../Data/mising_values_phone.csv")
phone_mv # mv means mising values

phone_mv['ram_gb'].mean()

# phone_mv['ram_gb'].fillna(phone_mv['ram_gb'].mean()) # this not change original file
# phone_mv['ram_gb'].fillna(phone_mv['ram_gb'].mean(), inplace = True)

# Another and safe way

phone_mv['ram_gb'] = phone_mv['ram_gb'].fillna(phone_mv['ram_gb'].mean())
```

```python
phone_mv = phone_mv.dropna() # drop all rows which contain missing values
phone_mv # now we have only one option to get back our data with NaN values (importing it again)
 
```

Remove and Add Data

```python
colour = pd.Series(["Black","Blue"], index = [2,6])

phone_mv['Colour'] = colour
phone_mv

# difference btw list and Series()

variants = [2,3] # if no of columns in data != list entries then it cause but Series() add NaN values in empty place.
phone_mv['variants'] = variants
phone_mv


phone_mv['product'] = phone_mv['variants'] * phone_mv['ram_gb']
phone_mv


phone_mv['new'] = 3
phone_mv

phone_mv['bool'] = True
phone_mv


phone_mv['new'] = 3
phone_mv.drop('new', axis = 1, inplace = True) 

# Axis = 1 for column and Axis = 0 for row

phone_mv
```

Data Manipulating :


```python
import pandas as pd
phone = pd.read_csv("../Data/phone.csv")
phone.sample(frac=0.5) # 50 % shufling
phone.sample(frac = 0) # 0% shufling
phone.sample(frac = 1) # 100% shufling
phone.reset_index(drop=True,inplace = True)
phone

# Now we want to change the storage_gb to storage_tb

# phone_mv['storage_gb'] = phone_mv['storage_gb'] / 1024
phone['storage_gb'] = phone['storage_gb'].apply(lambda x : x / 1024)
phone.columns = phone.columns.str.replace('storage_gb', 'storage_tb')   
phone
```
