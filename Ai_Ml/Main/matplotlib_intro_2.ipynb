# ***Pyplot vs Object Oriented Method***

```python
heart = pd.read_csv("/content/heart.csv")
heart
```

```python
heart_age = heart[heart['age'] < 50]
heart_age
```

```python
# pyplot method

heart_age.plot(kind = "scatter", x = 'age', y = 'chol', c = 'target');
```

```python
# Object Oriented approch

# fig,sub_plot = plt.subplots()
heart_age.plot(kind = 'scatter', x = 'age', y = 'chol', c = 'target'), sub_plot == sub_plot;
```

***Entire Life Circle of OO method***

```python
# create figure and graph area

fig, ax = plt.subplots()
scatter_plot = ax.scatter(x = heart_age['age'], y = heart_age['chol'], c = heart_age['target'])

# Customization :

ax.set(title='Scatter Graph', xlabel='Age', ylabel='Cholestrol')

# set lagend

ax.legend(*scatter_plot.legend_elements(), title = 'Target')

scatter_plot;
```

```python
# Advance

# create plot and graph area

fig, (ax1,ax2) = plt.subplots(figsize = (10,10), ncols = 1, nrows = 2, sharex=True) # share x axis

#  also change color there
# add data to plot 1

scatter_plot = ax1.scatter(x = heart_age['age'], y = heart_age['chol'], c = heart_age['target'], cmap = 'winter')

# costumize

ax1.set(title='Scatter Graph', ylabel='Cholestrol')

# lagend

ax1.legend(*scatter_plot.legend_elements(), title = "target")

# add mean line

ax1.axhline(heart_age['chol'].mean(), color = 'green', linestyle = '--')

plt.plot()

# now add new feature : change data color
# add data to plot 2

scatter_plot = ax2.scatter(x = heart_age['age'], y = heart_age['trestbps'], c = heart_age['target'], cmap = 'Dark2')

# costumize

ax2.set(title='Scatter Graph', xlabel='Age', ylabel='trestbps')

# lagend

ax2.legend(*scatter_plot.legend_elements(), title = "target")

# Add more features such as mean line in plot2

ax2.axvline(heart_age['age'].mean(), color = 'red', linestyle = '--')

plt.plot()

# apply universal style on plots (after this cell)

plt.style.use('dark_background')

# naming entire figure

fig.suptitle("Patient Data ", fontsize = 32,fontweight = 'bold');

# saving the figure/data

fig.savefig('patients_data.png')
```
